```sql
-- COLUMN LEVEL SECURITY: SI EL USUARIO NO HACE PARTEL DEL GRUPO DE ADMINS, NO PUEDE VISUALIZAR LA COLUMNA CUSTOMER NAME, MIENTRAS QUE COMO ADMIN PUEDE VER TODO. 
CREATE OR REPLACE VIEW bronze_raw_dynamic_vw AS 
SELECT customer_id, 
CASE WHEN is_account_group_member('admins') THEN customer_name ELSE 'Not Access' END AS customer_name, 
customer_address
FROM epuma_cust_bronze_raw;
```

```sql
SELECT *
FROM bronze_raw_dynamic_vw;
```

```sql
-- ROW LEVEL SECURITY: SI EL USUARIO NO HACE PARTEL DEL GRUPO DE ADMINS, SOLO PUEDE VISUALIZAR LAS FILAS DONDE EL CUSTOMER ID COMIENZA POR 4, MIENTRAS QUE COMO ADMIN PUEDE VER TODO. 
--NOTA: PARA FILTRAR A NIVEL DE COLUMNAS UTILIZAMOS CASE WHEN Y PARA FILTRAR A NIVEL DE FILAS UTILIZAMOS WHERE STATEMENT.
CREATE OR REPLACE VIEW bronze_raw_dynamic_vw AS 
SELECT *
FROM epuma_cust_bronze_raw
WHERE 
CASE WHEN is_account_group_member('admins') THEN TRUE ELSE customer_id LIKE '4%' END;

SELECT *
FROM bronze_raw_dynamic_vw;
```
