```sql
SELECT C.Name, 
       C.City, 
       G.GroupName, 
       G.CustomerNumber
FROM prodsqldwh.adls.companyinformation_dim C  
LEFT JOIN prodsqldwh.adls.customergroupmaster G 
ON C.CompanyCode = G.CompanyCode
ORDER BY StartDate DESC; 
```
