<b> Section 1: Data Discovery </b> 

<b> Unity Catalog. </b> 

Toda la información dentro de un Workspace en Databricks  es gobernada, administrada y se clasifica por medio del UC (Unity Catalog); El nivel más alto donde se encuentran almacenados todos los datos de un Workspace es el Unity Catalog Metastore y dentro de este también se encuentran configuraciones respecto de los controles de acceso al Workspace y solo tienen acceso los administradores; Después de este, los objetos de datos se organizan en 3 niveles jerárquicos o categorías que son: 
  - Catalog: Es el primer nivel o categoría de los objetos de datos y un Metastore puede tener múltiples Catalogs; Dicho objeto solo puede contener Schemas (o Bases de datos).
  - Schema: Es el segundo nivel o categoría de los objetos de datos; Equivalen a las bases de datos en diferentes RDBMS. 
  - Tablas, Funciones, Volúmenes, Vistas: Son el último nivel o categoría de los objetos de datos o de la información dentro de un Workspace.
    - Tablas: Las tablas son objetos de datos que almacenan información tabular de forma estructurada; Algunos de los tipos de tablas que se pueden almacenar en un Workspace son: 
      - Managed Tables: Son tablas creadas propiamente dentro del Workspace en el UC y por tanto, si se eliminan, no habrá ningún respaldo de la información. 
      - External Tables: Son tablas almacenadas dentro del Workspace en el UC pero donde la fuente de la información es externa (Snowflake, Sqlserver, Postgresql, Mysql, others)
      - Otros tipos de tablas son Streaming, Foreign y Feature Tables. 
    - Vistas: Son objetos de datos solo de consulta (read-only) y que se crean a partir de queries a una o múltiples tablas. 
    - Volúmenes: Son objetos de datos que permiten almacenar, organizar y acceder tanto a archivos de cualquier formato como csv, xlsx, txt, jpeg, png, json, xml, parquet como a datos estructurados, semi-estructurados y no estructurados; Esto con el fin de posteriormente convertir y almacenar los datos en forma tabular o utilizarlos para desarrollar modelos de machine learning y AI
    - Funciones: Son funciones de SQL que se puede almacenar dentro del workspace en el UC.

Por último, un Metastore puede estar asignado a múltiples Databricks Workspaces. 


<b> Principle of least privileges (PoLP):</b> 

En plataformas de administración de datos y donde se almacenan grandes cantidades de información como lo es el caso de Databricks, GCP, AWS, Snowflake y otros, siempre se aplica el principio de mínimos privilegios, refiriéndose a que siempre se deben otorgar los mínimos privilegios que sean posibles para que un usuario puede llevar a cabo una tarea específica o realizar sus labores, con el fin de evitar mal uso o accesos no autorizados.

<b> Identidades o principales:</b> 

En Databricks existen 3 tipos de identidades o principales que son: 
  - Users: Personas que tienen accesos y privilegios para hacer uso de la plataforma según sus labores. 
  - Service Principals: No corresponden a personas, y pueden ser conexiones de acceso con herramientas de terceros o externas y también servicios de automatización que ejecutan comandos en horarios específicos. 
  - Accounts Groups: Son cuentas que pueden tener tanto users como Service Principales, y son utilizadas por los administradores del workspace para conceder accesos a múltiples cuentas a la vez. 

<b> Propiedad de los objetos de datos:</b> 

Dentro del UC de databricks la propiedad de los objetos de datos se hereda de forma descendente, es decir, el creador o dueño de un schema se podrá otorgar privilegios de las tablas dentro de dicho schema así este no las haya creado; De igual forma sucede con el creador o propietario de un catágolo, que se podrá otorgar o conceder privilegios de acceso para los schemas y tablas dentro de dicho catálogo.

En general para los objetos de datos en el UC, el creador de dicho objeto puede otorgar privilegios y/o conceder la propiedad del mismo a otro principal o identidad, mientras que para objetos como Notebooks y Queries, el dueño solo puede ser modificado por el admin del workspace. 

<b>Objetos de datos Vs Objetos NO de datos:</b>
  - **Objetos de datos:** Se encuentran contenidos y/o organizados dentro del UC y corresponden a los Catálogos, Schemas y tablas, vistas, volúmenes, funciones. 
    - Tipos de privilegios o permisos: Como dueño o propietario de un objeto de datos se pueden otorgar o conceder los siguientes. 
      - Read: Execute, Select, Read Volume.
      - Edit: Modify, Refresh, Write Volume.
      - Create: Create schema, create table, create volume, create function, create materialized view, create model
      - Metadata: Apply tag, browse
      - Prerequisite: Use schema, use catalog
  - **Objetos NO de datos:** Notebooks, Files(texto, imagen),  Queries, Dashboards, Genies, Alertas.  
    - Tipos de privilegios o permisos: Como dueño o propietario de un objeto NO de datos se pueden otorgar o conceder los siguientes.
      - Can view, Can Read: ver o leer. 
      - Can Run: Puede ejecutar dicho objeto y refrescarlo para obtener info. actual. 
      - Can Edit: Puede modificar, agregar o quitar elementos del objeto. 
      - Can Manage: Puede conceder u otorgar privilegios o permisos a otros usuarios o principales; También puede eliminar el objeto. 








<b> Section 2: Data Ingestion (Data Engineering) </b> 

<b> Data Ingestion </b> 

Databricks utiliza Delta Lake para la ingestión, ingreso y almacenamiento de datos dentro de la plataforma; Delta Lake es un protocolo de código abierto, que permite la lectura y escritura de archivos para almacenarlos en la nube. Con el uso de dicho protocolo se busca almacenar la información por medio de Delta Tables, que son tablas de formato abierto que soportan arquitectura de Lakehouse. 

Dentro del Delta Lake, todos los archivos e información es almacenada en Delta Tables; Los datos e información dentro de las tablas, se almacenan posteriormente en directorios dentro del Delta Lake en formato o extensión Parquet, que es el formato por defecto dentro del protocolo; En el mismo directorio, junto con las tablas se almacenan los Delta Logs, que son archivos que permiten llevar el registro y control de todas las transacciones/operaciones que se hacen en una Delta Table (Insert, Update, Delete), lo que hace que se pueda acceder a diferentes estados de las tablas junto con su información (viajar atrás en el tiempo).

Key Features of the Delta Lake: Algunas de las características son:
  - Las transacciones realizadas dentro del protocolo son ACID, es decir, cumplen con: 
    - Atomicity
    - Consistency
    - Isolation
    - Durability
  - Es un protocolo que soporte DML (Data Managmente Language), es decir, permite realizar operaciones como Insert, Update, Delete, Joins permitiendo la administración flexible de los datos. 
  - Permite viajar atrás en el tiempo para acceder a diferentes versiones y/o estados de las tablas junto con su información, actuando como respaldo y poder revertir cambios de ser necesario.  
  - Permite evolución de los Schemas y refuerza la integridad de los datos. 

<b> Data Transformation</b> 

Después de haber realizado la ingesta y el ingreso de los datos e información de diferentes fuentes (CSP's, Bases de datos, Archivos de diversos formatos, otros) dentro del protocolo de Delta Lake, estos deben ser procesados y transformados para que puedan ser consumidos por las diferentes personas en la organización; En las plataformas de almacenamiento en la nube, y que utilizan arquitecturas de Lakehouse, se recomienda utilizar la Medallion Architecture (Arq. de Medalla) donde ha medida que la información pasa de una capa a otra, mejora la calidad y estructura de los datos. Los tipos de Layers son: 
  - Bronze: Donde se almancenan los datos justo después de haber hecho la ingesta de diferentes fuentes de información; Los datos son crudos, sin ningún tipo de limpieza ni procesamiento; Es la base para el resto de capas. 
  - Silver: Se almacenan los datos de la capa anterior, después de haber hecho la limpieza, eliminar duplicados, filtrar la data, se da estructura según los requerimientos del negocio y hacer cruces o Joins con otras tablas; Se consideran las tablas donde está la información más confiable del negocio (Single Source Truth).  
  - Gold: Los datos están totalmente procesados y listos para usarse, y la información está disponible para realizar tareas de BI y reportería. 


<b> Data Ingestion: Additional importing methods </b>

A parte de los métodos que normalmente utilizan los ingenieros de datos para la creación de Pipelines y hacer la ingesta de información dentro de Databricks, existen los siguientes métodos adicionales: 
  - Importar archivos por medio del UI: Directamente desde el Catalog explorer crear una tabla e indicar el archivo a partir del cual se quiere crear la tabla. 
  - read_files function: Después de almacenar el(los) archivo(s) dentro de un volumen en Databricks, se puede utilizar dicha función para importar los datos dentro de la tabla. 
  - COPY INTO Statement: Después de almacenar el(los) archivo(s) dentro de un volumen en Databricks, se puede hacer uso del COPY INTO statement para importar los datos dentro de la tabla. 


<b> Section 3: SQL Execution </b>

Dentro de la plataforma se integra Databricks SQL (que hace uso de AI y uso de lenguaje natural para la edición y creación de queries) y SQL Warehouse que corresponde a la capacidad de procesamiento dedicada a ejecutar SQL queries. Existen 3 tipos de Warehouses: 
  - Classic: Únicamente permite hacer uso de Photon, que es un Vectorized query engine integrado en Databricks que permite optimizar el uso de recursos en diferentes workloads, y acelerar la ejecucion de queries. 
  - Pro: Hace uso de Photon y Predictive IO. 
  - Serverless: Hace uso de las herramientas anteriores e integra Intelligent Workloads Managment (IWM, que permite asignar de manera de dinámica los recursos necesarios a los workdloads, en vez de tener Thresholds estáticos que pueden conllevar a capacidad ociosa); Con este tipo de SQL Warehouse o capacidad de procesamiento, la administración del uso de recursos, patching (reparación de errores) y optimizaciones de desempeño todo esta a cargo o es manejado por Databricks. 

  <b> Tipos de vistas: </b>
En SQL podemos utilizar las vistas que son objetos de datos de consulta que almacenan información de queries realizados a una o varias tablas; Los tipos de vistas son los siguientes: 
  - Standard View: Se almacena dentro del Schema del UC. 
  - Temporary View: Solo existe en la sesión actual de trabajo, y no ocupa almacenamiento dentro del Schema del UC. 
  - Materialized View: Tipo de vista que solo actualiza la información cuando se REFRESCA de forma manual o programada; Ocupa almacenamiento dentro del UC.
  - Dynamic View: Vista que permite de forma dinámica limitar el acceso de ciertos usuarios a la información o lo que se conoce como 'RLS' y 'CLS'

  


<b> Section 4: SQL Analysis: </b>

Las herramientas como Photon, Predictive I/O y IWM permiten optimizar la ejecución de SQL queries dentro de la plataforma de DB; Así mismo desde el Query Profile, podemos ver toda la información relacionada con la ejecución de queries, como las acciones realizadas y el tiempo consumido. 

Para el caso de Serverless SQL Warehouse, con el uso de IWM que se apalanca en el uso de AI y Machine Learning, se logra escalar de forma rápida recursos o clusters (Unidades de procesamiento) a Workloads o ejecución de Queries que lo requieran, y así mismo, reducir rápidamente los recursos asignados o clusters que ya no son utilizados para evitar capacidad ociosa (No se utilizan Clusters Always-on), mejorando la relación Costo/Rendimiento.
