
<b> Databricks: Lakehouse's architecture </b>

Databricks está creado bajo la arquitectura de Lakehouse, que es la combinación entre un Data Warehouse y un DataLake; Un Data Warehouse, permite el almacenamiento de datos y/o información tabular y estructurada y solo permite formatos de datos específicos, por lo que los datos almancenados en una bodega de datos son utilizados principalmente para tareas de reportería y BI; Por otro lado, un DataLake permite el almacenamiento de grandes volúmenes de datos, de cualquier formato, y tanto de datos estructurados como no estructurados, por lo cual se utiliza más para modelos de AI y Machine Learning; Databricks al tener una arquitectura híbrida de Lakehouse obtiene los beneficios de ambas arquitecturas de datos en una sola plataforma.  

<b> Schedule Queries: </b> 

Para agendar/programar un Query desde el SQL Editor, en la parte superior derecha se encuentra el botón Schedule que permite especificar la frecuencia y el rango de tiempo en que queremos que se ejecute de forma programada un Query; Si un Query programado no se ejecuta, lo más probable es que haya un error en el código de la consulta; Así mismo, los permisos de un Query no están relacionados con los permisos de un Query/programado agendado, es decir, que si un usuario o principal tiene acceso a un Query pero no cuenta con acceso a la ejecución programada/agendada de ese mismo Query no podrá visualizar los resultados después de que sean refrescados de forma automática.

- Los tipos de Caches o Memorias temporales que existen dentro de Databricks SQL para obtener información son las siguientes: 
  - UI Cache: Es la cache o memoria temporal que se ejecuta al recién abrir un Dashboard o Query, y trae el resultado de la última ejecución realizada; Se reemplaza cuando se ejecuta de nuevo un Query/consulta o después de pasados 7 días. 
  - Result Cache: 
    - Local Cache: Es la Cache o Memoria temporal que se ejecuta primero, dentro de la Result Cache; Solo existe mientras que el Cluster o unidad de procesamiento está activa, ya que si se detiene o reinicia el Cluster se elimina la información; Se utiliza para la ejecución de Queries repetitivos, evitando recalcular la misma información y tiene una duración de 24 hrs
    - Remote Cache: Es la Cache o Memoria temporal que se ejecuta si la Local Cache no cuenta con información; A diferencia de la memoria temporal anterior, la Remote Cache persiste aún después de que el cluster se detenga o reinicie y se almacena como información del Workspace; Persiste por 24 hrs.
  - Disk Cache: Se utiliza para lecturas rápidas o aceleradas de información. 
- El Liquid Clustering es un método de optimización Queries/consultas que permite ordenar las variables o columnas de una tabla según la frecuencia con que son utilizadas.
