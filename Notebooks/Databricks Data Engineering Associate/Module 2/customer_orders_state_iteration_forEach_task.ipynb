```python
# for each task iterating based on state from customer orders silver

# Reading and transforming data from master customers_orders_silver table
## Tranforming order data from unix timestamp to date and adding is_large_order column

state = dbutils.widgets.get('state')

df = spark.sql(
    f'''
    SELECT
        *,
        TO_DATE(
            FROM_UNIXTIME(order_datetime)
        ) AS order_date,
        CASE
            WHEN number_of_line_items > 2 THEN TRUE
            ELSE FALSE
        END AS is_large_order
    FROM de_module_2.jobs.customer_orders_silver
    WHERE state = '{state}'
    '''
    )

# Save to UC table, one per state specified in the loop:
table_name = f"de_module_2.jobs.customers_orders_{state.replace(' ', '_').lower()}_silver"
df.write.mode("overwrite").saveAsTable(table_name)
```
