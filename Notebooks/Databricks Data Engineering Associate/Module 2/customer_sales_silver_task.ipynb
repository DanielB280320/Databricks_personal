```python
%sql
CREATE OR REPLACE TABLE de_module_2.jobs.customer_sales_silver AS
SELECT
    CB.customer_id,
    CB.customer_name,
    CB.region,
    CB.state,
    CB.city,
    CB.ship_to_address,
    CB.valid_from,
    CB.valid_to,
    SB.order_date,
    CB.units_purchased,
    SB.product,
    SB.product_category,
    SB.product_name,
    SB.total_price

FROM de_module_2.jobs.customer_bronze CB
LEFT JOIN de_module_2.jobs.sales_bronze SB
    ON CB.customer_id = SB.customer_id
```

```python
customer_sales_df = spark.sql(
    '''
    SELECT
        *
    FROM de_module_2.jobs.customer_sales_silver
    '''
)
```

```python
duplicates_validation = customer_sales_df.count() > customer_sales_df.dropDuplicates().count()

# Setting a dynamic value
dbutils.jobs.taskValues.set(
    key= 'has_duplicates',
    value= duplicates_validation
)

duplicates_validation
```
