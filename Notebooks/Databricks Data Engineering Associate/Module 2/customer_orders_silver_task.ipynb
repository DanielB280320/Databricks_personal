```python
%sql
CREATE OR REPLACE TABLE de_module_2.jobs.customer_orders_silver AS
SELECT
    CB.customer_id,
    CB.customer_name,
    CB.region,
    CB.state,
    CB.city,
    CB.ship_to_address,
    CB.valid_from,
    CB.valid_to,
    CB.units_purchased,
    OB.clicked_items,
    OB.number_of_line_items,
    OB.order_datetime,
    OB.order_number,
    OB.promo_info

FROM de_module_2.jobs.customer_bronze CB
LEFT JOIN de_module_2.jobs.orders_bronze OB
    ON CB.customer_id = OB.customer_id
```

```python
customer_orders_df = spark.sql(
    '''
    SELECT
        *
    FROM de_module_2.jobs.customer_orders_silver
    '''
)
```

```python
duplicates_validation = customer_orders_df.count() > customer_orders_df.dropDuplicates().count()

dbutils.jobs.taskValues.set(
    key= 'has_duplicates',
    value= duplicates_validation
)

duplicates_validation
```
