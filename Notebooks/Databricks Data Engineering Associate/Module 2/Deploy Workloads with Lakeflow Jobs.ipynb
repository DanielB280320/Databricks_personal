# <b> Lakeflow Jobs Core Components </b>

## Overview

## Learning objectives

2. Identify and describe the task types and configuration options available in Lakeflow Jobs
3. Explain how Directed Acyclic Graphs (DAGs) enable task orchestration and workflow patterns
4. Learn how to create and configure jobs using the Lakeflow Jobs UI

## A. Building blocks of Lakeflow Jobs

### A1. Job and task

- **Job**: the primary resource for scheduling, coordinating and running operations (data processing, ETL, analytics, machine learning) in Databricks. Think of it as the container that holds the whole workflow
- **Task**: a single unit of work inside a job that runs a specific workload (notebook, script, query and more). Tasks do the actual work
- Hierarchical relationship: each job consists of one or more tasks

### A2. Task types

A job can mix many task types: Databricks Notebooks (any supported language), Spark Declarative Pipelines, SQL files/queries, Python scripts, Python wheels (packaged code), AI/BI dashboards, Spark Submit (legacy Spark apps), dbt, Java JAR files and Power BI. This variety lets you orchestrate virtually any workload from one job.

### A3. Task configuration options

- Options depend on the selected task type
- Common options: path to the code, dependent libraries, parameters, notifications and retry policies
- Dependencies are set per task with **Depends on** and **Run if dependencies** (for example, All succeeded)

### A4. Notebook task vs SQL task

- **Notebook task**: task name, type, source (for example Workspace), path, compute (cluster configuration), dependent libraries, parameters (UI or JSON) and notifications
- **SQL task**: task name, type, SQL task (for example Query), SQL query, SQL warehouse, parameters, notifications and retries
- The UI adapts to the task type and shows only the options that apply

### A5. Languages supported

Python, SQL, Scala, R and Java (through JAR files). Teams can keep their preferred languages and existing code assets.

### A6. Jobs orchestration: control flow, triggers and compute

- **Control flow between tasks**: sequential, parallel, conditional (if/else), Run Job (modular design) and For each (iterative processing)
- **Triggers**: manual (on-demand), scheduled (cron), API (programmatic), file arrival (event-driven), table (data change events) and continuous (streaming workloads)
- **Compute**: supports all of these orchestration patterns

### A7. Compute options

| Compute | Best for | Key points |
|---|---|---|
| Interactive (all-purpose) clusters | Development, ad-hoc analysis and exploration | Shared by multiple users. Should not be used in production because they are not cost-efficient |
| Job clusters | Production-grade and operational use cases | About 50% cheaper because they terminate when the job ends. Subject to cloud provider start-up times. The same cluster can be reused across tasks for better price performance |
| Serverless | Simple, fast, reliable, cost-efficient workloads | Fully managed, operationally simpler and more reliable. Faster clusters and auto-scaling, out-of-the-box performance optimizations and lower overall TCO |
| SQL warehouse | SQL queries, dashboards and BI | Serverless by default. High concurrency and autoscaling via Intelligent Workload Management. Auto-start/auto-stop, adjustable cluster size and max clusters help control costs |

### A8. Serverless performance mode

The **Performance optimized** setting on the job details page lets you choose between lower cost and faster execution. It applies only to tasks that use serverless compute.

- **Off (standard mode)**: focuses on cost-efficiency, longer startup time (typically 4-6 minutes), best for non-urgent workloads with flexible timing
- **On (optimized mode)**: faster job startup and execution, ideal for time-sensitive workloads

### A9. Selecting compute

- A job has one or more tasks and each task can be assigned its own compute resource
- Tasks in the same job can share the same compute or use different compute as required
- Example: Task-1 on an all-purpose cluster, Task-2 on serverless and Task-3 on a job cluster. This flexibility lets you optimize performance and cost per task

## B. Task orchestration

### B1. What is a DAG?

A DAG is a conceptual representation of a series of activities, including data processing flows.

- **Directed**: unambiguous direction for each edge
- **Acyclic**: contains no cycles
- **Graph**: a collection of vertices (tasks) connected by edges (dependencies)

Example: Task-1 leads to both Task-2 and Task-3, which represents the flow of execution.

### B2. Task orchestration overview

- Databricks Jobs supports task orchestration: running multiple tasks as a Directed Acyclic Graph (DAG)
- You define the order of execution by configuring task dependencies, which creates the DAG
- Tasks can be orchestrated through the Databricks UI, API, SDK or Declarative Automation Bundles
- Diagram example: Task-2 and Task-3 depend on Task-1, and Task-4 depends on Task-2 and Task-3

### B3. Common workload patterns

| Pattern | Shape | Typical use |
|---|---|---|
| Sequence | One task after another | Data transformation, processing and cleaning. Bronze/silver/gold tables (medallion architecture) |
| Funnel | Many tasks feed one task | Multiple data sources. Data collection and consolidation |
| Fan-out (star) | One task feeds many tasks | Single data source. Data ingestion and distribution to multiple downstream systems |

## C. Conclusion

- Lakeflow Jobs combines jobs, tasks, options, languages, triggers, control flows and compute for flexible orchestration
- It uses DAG-based orchestration to define clear task dependencies
- Tasks can be arranged in sequence, funnel or fan-out patterns to build reliable data workflows

## Quick recall

- Job = container and primary resource. Task = single unit of work. A job has one or more tasks
- Task types: notebooks, Spark Declarative Pipelines, SQL, Python scripts, Python wheels, AI/BI dashboards, Spark Submit, dbt, JAR, Power BI
- Languages: Python, SQL, Scala, R and Java (JAR)
- Control flow: sequential, parallel, if/else, Run Job, For each
- Triggers: manual, scheduled (cron), API, file arrival, table, continuous
- Compute: interactive (dev only), job clusters (about 50% cheaper, production), serverless (fully managed, lowest TCO), SQL warehouse (SQL and BI)
- Performance optimized applies only to serverless tasks. Off means 4-6 minutes startup, On means faster startup
- Patterns: sequence, funnel, fan-out

# Creating and Scheduling Jobs

*Course: Deploy Workloads with Lakeflow Jobs (Data Engineer Learning Plan)*

## Overview

How to fine-tune Lakeflow Jobs tasks with parameters, dynamic values, notifications and retries, and how to automate runs with schedules and event-based triggers.

## Learning objectives

1. Configure task parameters, task values, dynamic value references, notifications and retry policies
2. Set up and manage scheduling options: scheduled, file arrival, table update and continuous triggers
3. Automate job runs with triggers for both time-based and event-driven workflows
4. Explore the scheduling options in the Lakeflow Jobs UI through the hands-on demo

## A. Common task configuration options

### A1. Three main categories

- **Parameters and dynamic value references**: make tasks reusable and adaptable, because a task can behave differently depending on its inputs or runtime context. Available at task and job level
- **Retries**: the first line of defense against transient failures. Set at job and task level, so each part of the workflow can use a strategy that fits how critical it is
- **Notification alerts**: keep the team informed so they can react fast. Also set at job and task level

### A2. Parameter types

| Type | Defined at | How it works |
|---|---|---|
| Task parameters | Individual task | Key-value pairs (or a JSON array) passed into that one task for fine-grained control. They enable conditional execution, looping and passing context between tasks |
| Job parameters | Job | Sent automatically to every task in the job. Act as shared defaults and can be overridden when a run is triggered (for example another environment, date range or processing mode) |

- **Precedence rule**: when the same key exists in both, the **job parameter wins**. Task parameters give per-task defaults, job parameters give centralized runtime overrides
- Typical uses of task parameters: choose which branches run based on data conditions or business rules (conditional execution), control iteration counts or the array a for each loop walks through (looping), and set values that downstream tasks read (context passing)

### A3. Setting and reading parameters

- **Job parameters**: open the Parameters section of the job and add key-value pairs. Common examples are catalog name, schema name, environment and processing date
- **Task parameters**: open the task configuration (next to task name, type and path) and add the pairs there. Good for task-specific paths, processing options or overrides
- **Reading them in a notebook task**: `dbutils.widgets.get("param_name")` returns both kinds, and if a key exists at both levels you receive the job value
- How you read parameters depends on the task type (notebook, SQL, Python wheel, JAR), so check the docs for each one
- Lesson example: parameters `catalog_name` = dbacademy and `schema_name` = ops, read in the notebook with `dbutils.widgets.get("catalog_name")` and `dbutils.widgets.get("schema_name")`

### A4. Task values (set at runtime)

- Task values are key-value pairs that a task creates **while it runs** and shares with later tasks. Parameters, by contrast, are fixed before execution starts
- Set a value: `dbutils.jobs.taskValues.set(key="catalog_name", value="dbacademy")`
- Read it from an upstream task: `dbutils.jobs.taskValues.get(taskKey="task-name", key="catalog_name")`
- This builds data-based dependencies between tasks, not just execution order
- Good for sharing record counts for quality checks, generated file paths, processing statistics, or criteria for conditional logic

### A5. Dynamic value references

- Placeholders written with double curly braces that Databricks resolves at runtime from the job and task context (task outputs, parameters, task values, run metadata)
- The parameter value field in the UI offers autocomplete for the available references, including fields of upstream tasks such as their result state or error code

| Reference | What it gives you |
|---|---|
| `{{job.start_time.day}}` | Day the job started, for date-based processing |
| `{{job.run_id}}` | Unique run identifier for tracking and logging |
| `{{job.parameters.environment}}` | Value of a job-level parameter |
| `{{task.name}}` | Task name, useful for logging and building dynamic paths |
| `{{task.retry_count}}` | Retry attempt number, useful for debugging |
| `{{tasks.data-validation.values.record_count}}` | Task value produced by an upstream task |
| `{{tasks.file-processor.values.output_path}}` | Path generated by another task |

These references let a workflow adapt to different environments, data volumes and upstream results.

### A6. Notification alerts

- **Job-level**: configured in the right-side pane of the Job details. The alert is sent once the whole job completes successfully. Useful for people waiting on a complete workflow, such as business users expecting a daily report or downstream systems that consume the job output
- **Task-level**: every task can have its own configuration, so alerting can be granular. Example: immediate alerts for data validation failures but only a summary for routine cleanup tasks
- Design the strategy around operational needs: who needs to know what, when, and what they can do about it

### A7. Designing notification strategies

- **Destinations**: email, Microsoft Teams, PagerDuty, Slack and webhook. System destinations are set up by the workspace administrator
- **Events**: start, success, failure, duration warning and streaming backlog
- **Late jobs**: duration threshold warnings and timeout alerts catch slowdowns early. Streaming backlog warns when a streaming workload is falling behind
- **Mute options**: skipped runs, canceled runs, and everything until the last retry
- **Webhook**: custom integration with third-party APIs and custom decision logic
- Different tasks can notify different teams. For example, ingestion tasks alert data engineering while reporting tasks notify business stakeholders. Developers may prefer Slack while operations prefer PagerDuty

### A8. Retry policy

- Defines when and how many times a failed run is retried
- In the UI you pick the maximum number of retries (the total attempts are one more) and the wait time between retries. The dropdown offers 5, 6, 7, 8, 10, 20 or 30 retries, or unlimited, and there is also a retry on timeout option
- The task retry setting in the lesson example reads: immediately, at most 3 times (4 total attempts)
- Factors to weigh when designing a policy: failure type (transient network issues deserve immediate retries, data quality problems usually do not), resource impact (aggressive retries of heavy tasks can cause cluster contention), downstream dependencies, and business SLAs that limit the retry window

## B. Job schedules and triggers

A trigger is a rule that starts a job run automatically when a condition is met or on a schedule. Triggers remove manual steps, so jobs run around the clock and do not depend on someone being available. Each trigger type has its own reliability characteristics, so pick the one that fits the use case.

### B1. Trigger types

| Trigger | What it does | Key points |
|---|---|---|
| Scheduled | Runs the job at fixed times or intervals | Simple mode (hourly, daily, weekly, monthly) or Advanced mode with cron syntax. Always set the right time zone. Status can be Active or Paused. Examples: daily ETL in the morning, weekly executive report, monthly KPIs |
| File arrival | Starts the job when new files land in a monitored location | Event-driven. Supports AWS S3, Azure storage, Google Cloud Storage and Databricks Volumes (Unity Catalog volumes or external locations). Watches paths of up to 10,000 files. Advanced settings: minimum time between triggers and wait after last change. Fits irregular ingestion such as partner feeds, IoT files, financial data or logs. File pattern matching can skip temporary or incomplete uploads |
| Continuous | Starts a new run as soon as the previous one finishes or fails | Databricks manages restart and retry automatically. Built for streaming workloads such as real-time analytics, fraud detection, IoT and change data capture. Needs a different monitoring approach because the job is meant to run indefinitely |
| Manual | Runs on demand (trigger type None) | Run now, or Run now with different settings to override parameters. Can also be started from the API, CLI, SDK or Declarative Automation Bundles. Best for ad hoc runs, debugging, testing and one-off backfills. Can be combined with other trigger types later |
| Table update | Starts the job when monitored tables change | Detects insert, update, delete and merge. Replaces cron-style schedules with real-time orchestration, so data is fresher and less compute is wasted |

### B2. Table update trigger configuration

1. **Select the trigger type**: choose Table update
2. **Add source tables**: up to 10 tables per trigger. Works with Unity Catalog managed Delta or Iceberg tables, external UC tables, materialized views and streaming tables
3. **Define the condition**: *any table updated* fires as soon as the first monitored table changes, *all tables updated* fires only after every selected table has been updated
4. **Advanced options (optional)**:
   - **Minimum time between triggers**: enforces a gap between consecutive runs so a frequently updated table does not cause many runs in quick succession
   - **Wait after last change**: delays the run until a set time has passed since the latest update, so a multi-batch delivery has fully landed before the job starts

The UI also suggests adding failure notifications in case the trigger evaluation fails.

## C. Conclusion

- Parameters, dynamic value references, notifications and retries make workflows reusable, adaptable, monitorable and resilient
- Triggers start runs automatically from schedules or conditions: scheduled, continuous, file arrival, manual and table update
- The table update trigger enables real-time orchestration by watching tables for changes instead of relying on time-based scheduling

## Quick recall

- Job parameter beats task parameter when the key is the same
- Parameters are fixed before the run. Task values are created during the run with `taskValues.set` and read with `taskValues.get`
- Notebook tasks read parameters with `dbutils.widgets.get`
- Dynamic value references use double curly braces
- Notification destinations: email, Teams, PagerDuty, Slack, webhook. Events: start, success, failure, duration warning, streaming backlog
- Retries: set max retries and wait time, optional retry on timeout
- Trigger types: scheduled, continuous, file arrival, manual, table update
- File arrival watches up to 10,000 files. Table update watches up to 10 tables, with any or all conditions
