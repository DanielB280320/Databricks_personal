# <b> Introduction to Data Engineering in Databricks </b>
*Course: Deploy Workloads with Lakeflow Jobs (Data Engineer Learning Plan)*

## Overview
Databricks pairs optimized storage with unified governance (Unity Catalog) and Lakeflow's integrated tools for ingestion, transformation and orchestration. This lesson focuses on **Lakeflow Jobs**, the orchestration piece, and why a native orchestrator beats external ones.

## Learning objectives
1. Know the data engineering building blocks on Databricks: Connect, Spark Declarative Pipelines (SDP) and Jobs
2. Explain what Lakeflow Jobs is and its main benefits
3. Identify Lakeflow Jobs capabilities and use cases

## A. Data engineering platform on Databricks
Layers, top to bottom:
- **Lakeflow**: Connect (efficient ingestion connectors), Spark Declarative Pipelines (accelerated ETL, previously DLT) and Jobs (reliable orchestration for analytics and AI, previously Workflows)
- **Processing engine**: Apache Spark plus Structured Streaming (Photon)
- **Unified governance**: Unity Catalog
- **Optimized storage**: Delta Lake, Parquet and Iceberg

Lakeflow Jobs can orchestrate every kind of workload on Databricks: notebooks, SQL queries, dashboards, pipelines and more.

## B. What is Lakeflow Jobs?
### Orchestration options
Open source (Apache Airflow, Prefect, Dagster, dbt), cloud-native services (AWS, Azure, Google Cloud) or custom in-house frameworks. A typical workflow ingests sessions and clicks, joins them, featurizes and aggregates, analyzes and trains models, then feeds BI and warehousing, streaming, and data science/ML. The open question: which orchestration approach is best?

### Problems with external orchestrators
- Teams are less productive because the tools are hard for many practitioners to use
- Bad data quality lowers the value of downstream applications
- Higher cost of ownership and lower reliability: root cause is hard to find and the architecture gets complex
- Not unified with the Lakehouse, which creates integration challenges and data silos

### Lakeflow Jobs
Unified orchestration for data, analytics and AI on the Data Intelligence Platform.
- Key benefits: **simple authoring**, **actionable insights**, **proven reliability**
- Native to the platform, so it integrates with ingestion and transformation, the processing engine (Photon), governance (Unity Catalog), storage (Delta Lake), data warehousing and machine learning
- The whole workflow (sessions and clicks, join, featurize, aggregate, analyze, train) runs inside one platform, with no integration work against external tools

### Architecture
A central **Workflow** engine (a graph of tasks) surrounded by:

| Component | Role |
|---|---|
| Triggers | Scheduled (time-based), Continuous (always running), File arrival (event-driven), Table updates |
| Control flow | Manages task dependencies and execution order |
| Observability | Monitoring and troubleshooting |
| Compute | Runs the workload types: ETL, ML/AI and Analytics/BI |

## Conclusion
- Optimized storage, Unity Catalog governance and Lakeflow cover ingestion, transformation and orchestration
- Lakeflow Jobs orchestrates data, analytics and AI workloads natively, unifying workflows, triggers, compute, observability and control flow
- This reduces reliance on external orchestrators
